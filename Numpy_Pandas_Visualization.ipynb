# 25. Create a class Student with private attributes name and marks. Provide getter and setter methods. Prevent marks from being set below 0 or above 100.  

```python
class Student :
    def __init__(self, name, marks) :
        self.__name__ = name
        self.__marks__ = marks

    def get_name(self) :
        print(f"Name : {self.__name__}")

    def get_marks(self) :
        print(f"Marks : {self.__marks__}")

    def setter(self, marks) :
        if marks >= 0 and marks <= 100 : 
            self.__marks__ =  marks
        else :
            print("Marks must be between o and 100")
```

```python
s1 = Student("Tom", 85)
s1.get_name()
```

```python
s1.get_marks()
```

# 26. Print first N prime numbers efficiently.

```python
start = int(input("Enter the starting point : "))
end = int(input("Enter the ending point : "))

for num in range(start, end + 1) :
    if num > 1 :
        for i in range(2, int(num ** 0.5) + 1) :
            if num % i == 0 :
                break
            else :
                print(num, end = " ")
```

# 27. Load the "tips" dataset from seaborn (2)
# i) Show the outliers of the numeric columns (3)#
# ii) Show the count of Male and Female Person in a plot (2)
# iii) Compare the day average of wise total bill (3)
# iv) Show the percentage proportion of days (2)

```python
import seaborn as sns
```

```python
df = sns.load_dataset("tips")
```

```python
df.head()
```

```python
# i) Show the outliers of the numeric columns 
import matplotlib.pyplot as plt
for col in df.columns :
    if df[col].dtype != "object" :
        sns.boxplot(df[col])
        plt.title(col)
        plt.show()
```

```python
# ii) Show the count of Male and Female Person in a plot
df["sex"].value_counts()
```

```python
sns.countplot(data = df, x = "sex")
```

```python
# iii) Compare the day average of wise total bill
day_avg = df.groupby("day")["total_bill"].mean().reset_index()
```

```python
sns.barplot(data = day_avg, x = "day", y = "total_bill")
plt.title("Day wise average total bill")
plt.show()
```

```python
# iv) Show the percentage proportion of days
day_count = df["day"].value_counts()
plt.pie(day_count, labels = day_count.index, autopct = "%0.2f%%")
```
