# 1. Result of np.array([1,2,3]) + np.array([4,5,6])?
a. [5 7 9] b. [4 10 18] c. Error d. [1 2 3 4 5 6]

Answer :- [5, 7, 9]

# 2. Which function creates an array filled with ones?
a. np.ones() b. np.full() c. np.eye() d. np.identity()

Answer :- np.ones()

# 3. Which operation is element-wise in NumPy?
a. np.add() b. np.dot() c. np.matmul() d. np.cross()

Answer :- np.add()

# 4. Which function creates an identity matrix?
a. np.eye() b. np.zeros() c. np.full() d. np.reshape()

Answer :- np.eye()

# 5. Which function is used to create a NumPy array?
a. np.list() b. np.array() c. np.make() d. np.data()

Answer :- np.array()

# 6. What does np.random.randint(1,10) generate?
a. Float numbers b. Array of numbers c. Single random integer d.Sorted numbers

Answer :- Single random integer

# 7. Which slicing gives all rows and 2nd column?
a. a[2,:] b. a[:,1] c. a[1,:] d. a[:,:2]

Answer :- a[ : , 1]

# 8. What does np.choice() do?
a. Selects random elements b. Sorts array c. Reshapes array d. Creates identity matrix

Answer :- Selects random elements

# 9. What does np.flatten() do?
a. Converts array b. Reshapes array c. Converts to 1D array d. Deletes elements

Answer :- Converts to 1D array

# 10. What is the output of np.arange(5)?
a. [1 2 3 4 5] b. [0 1 2 3 4] c. [0 1 2 3 4 5] d. [5]

Answer :- [0 1 2 3 4]

# 11. What happens if reshape size doesn’t match elements?
a. Error b. Auto fill zeros c. Ignore extra values d.Repeats values

Answer :- Error

# 12. What is NumPy mainly used for?
a. Web development b. Numerical computing c. Game development d. Mobile apps

Answer :- Numerical computing

# 13. What does np.resize() do?
a. Changes array shape and repeats values b. Deletes elements c. Flattens array d. Shuffles array

Answer :- Changes array shape and repeats values

# 14. What does np.zeros((3,3)) create?
a. 3x3 matrix of ones b. 3x3 identity matrix c. 3x3 matrix of zeros d. Random matrix

Answer :- 3x3 matrix of zeros

# 15. What does np.eye(3,4) create?
a. 3x3 identity b. 4x4 identity c. 3x4 matrix with diagonal ones d. 4x3 matrix

Answer :- 3x4 matrix with diagonal ones

# 16. Which function shuffles an array in-place?
a. np.shuffle() b. np.random.shuffle() c. np.random.choice() d. np.sort()

Answer :- np.random.shuffle()

# 17. What does np.reshape(2,3) change?
a. Number of elements b. Array values c. Array shape d. Array type

Answer :- Array shape

# 18. Which function generates floating random numbers?
a. np.random.randint() b. np.random.uniform() c. np.random.choice() d. np.arange()

Answer :- np.random.uniform()

# 19. np.linspace(0,10,5) returns:
a. 5 random numbers b. Numbers with step size 5 c. 5 evenly spaced numbers d. Integers only

Answer :- 5 evenly spaced numbers
